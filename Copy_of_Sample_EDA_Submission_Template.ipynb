## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
# Write your code to make your dataset analysis ready

# 1. Dropping columns that have too many missing values or are not needed for basic analysis
cols_to_drop = ['imdb_id', 'tmdb_popularity', 'description']
titles_cleaned = titles_df.drop(columns=cols_to_drop)

# 2. Handling missing values in important columns
# Filling missing age ratings with 'Not Rated'
titles_cleaned['age_certification'] = titles_cleaned['age_certification'].fillna('Not Rated')

# Filling missing scores with the median value (so that the average isn't affected much)
titles_cleaned['imdb_score'] = titles_cleaned['imdb_score'].fillna(titles_cleaned['imdb_score'].median())
titles_cleaned['tmdb_score'] = titles_cleaned['tmdb_score'].fillna(titles_cleaned['tmdb_score'].median())

# 3. Handling the 'seasons' column (filling NaN with 0 for Movies)
titles_cleaned['seasons'] = titles_cleaned['seasons'].fillna(0)

# 4. Final check to see if null values are reduced
print("Missing values after cleaning:")
print(titles_cleaned.isnull().sum())

print("\nCleaned Dataset Shape:", titles_cleaned.shape)
````

### What all manipulations have you done and insights you found?

Data Manipulations Performed
To make the raw data useful for analysis, I performed the following steps:

1.Data Cleaning: Removed irrelevant columns like imdb_id and description to focus on analytical features.

2.Handling Missing Values: * Filled missing Age Certifications with 'Not Rated'.

3.Replaced null IMDb/TMDB scores with the median value to avoid losing data points.

4.Set Seasons to 0 for titles where the information was missing (mostly movies).

5.Data Type Casting: Ensured that release_year and runtime are treated as numerical values for calculation.

6.Consistency Check: Verified that there are no duplicate records to ensure the accuracy of our count-based visualizations.

````python
# Chart - 9 visualization code
# Chart - 9 visualization code
# Average IMDb Score by Top 10 Production Countries (Error Fixed)

# 1. Cleaning and exploding countries
df_country = titles_df[['production_countries', 'imdb_score']].copy()
df_country['production_countries'] = df_country['production_countries'].str.replace("[", "").str.replace("]", "").str.replace("'", "").str.split(',')

# Explode creates duplicate indices, so we reset it immediately
df_country = df_country.explode('production_countries')
df_country['production_countries'] = df_country['production_countries'].str.strip()

# Resetting index to avoid "Duplicate Labels" error
df_country = df_country.reset_index(drop=True)

# 2. Get Top 10 Countries by count
top_10_countries = df_country['production_countries'].value_counts().head(10).index

# 3. Filter for only those top 10 and calculate mean
top_country_scores = df_country[df_country['production_countries'].isin(top_10_countries)]
avg_scores = top_country_scores.groupby('production_countries')['imdb_score'].mean().sort_values(ascending=False)

# 4. Visualization
plt.figure(figsize=(12, 6))
sns.barplot(x=avg_scores.values, y=avg_scores.index, palette='viridis')

plt.title('Average IMDb Score by Top 10 Production Countries', fontsize=15, fontweight='bold')
plt.xlabel('Average IMDb Score', fontsize=12)
plt.ylabel('Country', fontsize=12)
plt.xlim(0, 10)

for i, v in enumerate(avg_scores.values):
    plt.text(v + 0.1, i, f'{v:.2f}', va='center', fontweight='bold')

plt.show()
````
